# 第6章 LRPT（Meteor-M）
気象衛星の記録から画像を再現します。HackRF で記録した Meteor-M N2-4 の `meteor5.iq` を `data/` に置いて実行してください。なければ、sigidwiki の LRPT のページで公開されている録音（WAV）を使います。

```python
import os, sys
from pathlib import Path
sys.path.insert(0, "..")  # code/ 直下のモジュールを読み込む
DATA = Path(os.environ.get("SDR_BOOK_DATA", "../data"))  # IQ データの置き場所
import numpy as np
import matplotlib.pyplot as plt
```

```python
import subprocess

iq = DATA / "meteor5.iq"  # HackRF で記録した N2-4（中心 137.65MHz、2MHz）
if iq.exists():
    cmd = [str(iq), "-f", "cs8", "-r", "2000000", "--offset", "250000", "--decim", "8", "-m", "oqpsk"]
else:  # sigidwiki の録音（古い N2、QPSK）
    cmd = [str(DATA / "LRPT_16-29-02_137900kHz.wav")]
subprocess.run([sys.executable, "../lrpt.py", *cmd, "-o", "lrpt_out"], check=True)
```

```python
from PIL import Image
for p in sorted(Path("lrpt_out").glob("*.png")):
    print(p)
plt.figure(figsize=(8, 8)); plt.imshow(Image.open(sorted(Path("lrpt_out").glob("rgb_*.png"))[0])); plt.axis("off"); plt.show()
```
