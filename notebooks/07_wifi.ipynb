# 第7章 Wi-Fi（IEEE 802.11a/g）
OFDMの同期、チャネル推定、QAMの復調、ビタビ復号を経て、ビーコンを読みます。

```python
import os, sys
from pathlib import Path
sys.path.insert(0, "..")  # code/ 直下のモジュールを読み込む
DATA = Path(os.environ.get("SDR_BOOK_DATA", "../data"))  # IQ データの置き場所
import numpy as np
import matplotlib.pyplot as plt
```

```python
import wifi

x = wifi.load_iq(DATA / "wifi_5g_ch120.iq")
bursts = wifi.detect_bursts(x)
print("bursts:", len(bursts))
```

## 最初に復号できたフレーム

```python
for s, e in bursts:
    burst = x[max(0, s - 40):e + 200]
    r = wifi.decode_ppdu(burst)
    if r is not None and wifi.check_fcs(r["psdu"]):
        break
print(r["mbps"], "Mbps", wifi.parse_frame(r["psdu"])["type"])
pts = r["points"]
plt.figure(figsize=(4, 4)); plt.plot(pts.real, pts.imag, ".", ms=2); plt.axis("equal"); plt.title("constellation"); plt.show()
```
