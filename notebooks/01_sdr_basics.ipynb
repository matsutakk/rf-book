# 第1章 無線通信とSDR
IQデータを読み込み、スペクトルとスペクトログラムを表示します。

```python
import os, sys
from pathlib import Path
sys.path.insert(0, "..")  # code/ 直下のモジュールを読み込む
DATA = Path(os.environ.get("SDR_BOOK_DATA", "../data"))  # IQ データの置き場所
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sdr_basics import load_iq, show_spectrum

fs, fc = 2_000_000, 1_090_000_000
x = load_iq(DATA / "adsb.iq", count=fs * 2)  # 先頭1秒分
print(len(x), x[:4])
```

IQ平面にサンプルを打ってみます。

```python
plt.figure(figsize=(4, 4))
plt.plot(x[:5000].real, x[:5000].imag, ".", ms=1)
plt.xlabel("I"); plt.ylabel("Q"); plt.axis("equal"); plt.show()
```

```python
show_spectrum(x, fs, fc)
```
